```python
from pathlib import Path
import pandas as pd
```

```python
csv_path = Path("../Datasets/Original/train3098itemPOLARITY.csv")
```

```python
df = pd.read_csv(csv_path, sep=";", encoding="utf-8-sig", on_bad_lines="skip")
```

```python
print("== Senti4SD dataset Inspection ==")
print("Total Records (Rows): ", len(df))
print("Total Columns: ", len(df.columns))
print("Column Names: ", df.columns.tolist())
print("\nMissing Values:\n", df.isna().sum())
```

```python
label_col = [
    c for c in df.columns if any(w in c.lower() for w in ["polarity", "sentiment"])
][0]
print(f"\nLabel Distribution ({label_col}):")
print(df[label_col].value_counts(dropna=False))
```
