```python
from pathlib import Path
import pandas as pd
```

```python
root = Path.cwd()

if not (root / "Datasets").exists():
    root = root.parent

csv_path = root / "Datasets" / "Original" / "train3098itemPOLARITY.csv"
```

```python
df = pd.read_csv(csv_path, sep=";", header=None, names=["ID", "Polarity", "Text"], encoding="utf-8-sig", dtype=str, keep_default_na=False)
```

```python
print("Total records:", len(df))
print("Columns:", df.columns.tolist())
print("\nFirst three records:")
print(df[["ID", "Polarity", "Text"]].head(3).to_string(index=False))
```

```python
print("Total records:",len(df))
```

```python
print("Unique IDs:", df["ID"].nunique())
print("Duplicate IDs:", df["ID"].duplicated().sum())
print("Exact duplicate rows:", df.duplicated().sum())
```

```python
normalised_text = (df["Text"].str.strip().str.casefold())
print("Duplicate texts:", normalised_text.duplicated().sum())
print("Blank text records:", df["Text"].str.strip().eq("").sum())
print("Blank IDs:", df["ID"].str.strip().ep("").sum())
```
